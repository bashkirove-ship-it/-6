```python
# Домашнее задание № 3

# Задание № 1

# Использую изученный на лекции метод json

import json

purchases = {}

with open('purchase_log.txt', 'r', encoding='utf-8') as f:
    for line in f:
        line = line.strip()
        if not line:
            continue
        data = json.loads(line)
        if data['user_id'] == 'user_id':
            continue
        purchases[data['user_id']] = data['category']

count = 0
for uid, category in purchases.items():
    print(f"{uid} '{category}'")
    count += 1
    if count == 2:
        break
```
